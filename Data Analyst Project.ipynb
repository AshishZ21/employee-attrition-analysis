```python
import pandas as pd
df=pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")
df.info()
```

```python
df.isnull().sum()
df.duplicated().sum()
```

```python
# ==========================================
# 1. Check Dataset Information
# ==========================================
print("Dataset Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
```

```python
# ==========================================
# 2. Check Missing Values
# ==========================================
print("\nMissing Values:")
print(df.isnull().sum())

# Total Missing Values
print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
# ==========================================
# 3. Check and Remove Duplicates
# ==========================================
duplicate_count = df.duplicated().sum()
print("\nNumber of Duplicate Rows:", duplicate_count)

# Remove duplicates
df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
# ==========================================
# 4. Check Data Types
# ==========================================
print("\nCurrent Data Types:")
print(df.dtypes)

# Convert categorical columns to category datatype
categorical_cols = [
    'Attrition',
    'BusinessTravel',
    'Department',
    'EducationField',
    'Gender',
    'JobRole',
    'MaritalStatus',
    'OverTime'
]

for col in categorical_cols:
    df[col] = df[col].astype('category')

# Verify updated data types
print("\nUpdated Data Types:")
print(df.dtypes)
```

```python
# ==========================================
# 5. Rename Columns for Better Readability
# ==========================================
df = df.rename(columns={
    'Age': 'Age',
    'Attrition': 'Attrition_Status',
    'BusinessTravel': 'Business_Travel',
    'DailyRate': 'Daily_Rate',
    'DistanceFromHome': 'Distance_From_Home',
    'EducationField': 'Education_Field',
    'EmployeeCount': 'Employee_Count',
    'EmployeeNumber': 'Employee_ID',
    'EnvironmentSatisfaction': 'Environment_Satisfaction',
    'HourlyRate': 'Hourly_Rate',
    'JobInvolvement': 'Job_Involvement',
    'JobLevel': 'Job_Level',
    'JobRole': 'Job_Role',
    'JobSatisfaction': 'Job_Satisfaction',
    'MaritalStatus': 'Marital_Status',
    'MonthlyIncome': 'Monthly_Income',
    'MonthlyRate': 'Monthly_Rate',
    'NumCompaniesWorked': 'Companies_Worked',
    'OverTime': 'Over_Time',
    'PercentSalaryHike': 'Salary_Hike_Percent',
    'PerformanceRating': 'Performance_Rating',
    'RelationshipSatisfaction': 'Relationship_Satisfaction',
    'StandardHours': 'Standard_Hours',
    'StockOptionLevel': 'Stock_Option_Level',
    'TotalWorkingYears': 'Total_Working_Years',
    'TrainingTimesLastYear': 'Training_Last_Year',
    'WorkLifeBalance': 'Work_Life_Balance',
    'YearsAtCompany': 'Years_At_Company',
    'YearsInCurrentRole': 'Years_In_Current_Role',
    'YearsSinceLastPromotion': 'Years_Since_Last_Promotion',
    'YearsWithCurrManager': 'Years_With_Current_Manager'
})
```

```python
# ==========================================
# 6. Final Verification
# ==========================================
print("\nUpdated Column Names:")
print(df.columns)

print("\nFinal Dataset Shape:", df.shape)

# ==========================================
# 7. Save Cleaned Dataset
# ==========================================
df.to_csv("HR_Analytics_Cleaned.csv", index=False)

print("\nCleaned dataset saved as 'HR_Analytics_Cleaned.csv'")
```

```python
# Calculate attrition count by job role
attrition_roles = df[df['Attrition_Status'] == 'Yes'] \
    .groupby('Job_Role') \
    .size() \
    .reset_index(name='Attrition_Count') \
    .sort_values(by='Attrition_Count', ascending=False)

print("Attrition by Job Role:")
print(attrition_roles)
```

```python
top_5_attrition = attrition_roles.head(5)

print("Top 5 Highest Attrition Roles:")
print(top_5_attrition)
```

```python
lowest_5_attrition = attrition_roles.tail(5)

print("Top 5 Lowest Attrition Roles:")
print(lowest_5_attrition)
```

```python
role_summary = pd.crosstab(
    df['Job_Role'],
    df['Attrition_Status']
)

role_summary['Attrition_Rate (%)'] = (
    role_summary['Yes'] /
    (role_summary['Yes'] + role_summary['No'])
) * 100

role_summary = role_summary.sort_values(
    by='Attrition_Rate (%)',
    ascending=False
)

print(role_summary)
```

```python
salary_analysis = df.groupby(
    'Attrition_Status'
)['Monthly_Income'].mean()

print(salary_analysis)
```

```python
salary_bins = [0, 3000, 6000, 10000, 20000]
salary_labels = [
    'Low Income',
    'Medium Income',
    'High Income',
    'Very High Income'
]

df['Salary_Band'] = pd.cut(
    df['Monthly_Income'],
    bins=salary_bins,
    labels=salary_labels
)

salary_attrition = pd.crosstab(
    df['Salary_Band'],
    df['Attrition_Status']
)

salary_attrition['Attrition_Rate (%)'] = (
    salary_attrition['Yes'] /
    (salary_attrition['Yes'] + salary_attrition['No'])
) * 100

print(salary_attrition)
```

```python
satisfaction_metrics = [
    'Job_Satisfaction',
    'Environment_Satisfaction',
    'Work_Life_Balance'
]

for col in satisfaction_metrics:

    summary = pd.crosstab(
        df[col],
        df['Attrition_Status']
    )

    summary['Attrition_Rate (%)'] = (
        summary['Yes'] /
        (summary['Yes'] + summary['No'])
    ) * 100

    print(f"\n{'='*50}")
    print(f"{col}")
    print(summary[['Attrition_Rate (%)']])
```
